```python
# Verifying Transformer training dataset

from pathlib import Path
import json
from collections import Counter

from src.config import get_settings

settings = get_settings()

print("===== TRANSFORMER TRAINING DATA =====")

data_dir = Path("data/processed")

for split in ["train", "val", "test"]:
    path = data_dir / f"{split}.jsonl"

    print(f"\n{split.upper()}:")
    print("Path:", path)
    print("Exists:", path.exists())

    assert path.exists(), f"Missing dataset: {path}"

    rows = []
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                rows.append(json.loads(line))

    print("Samples:", len(rows))
    print("Fields:", list(rows[0].keys()))

    intents = [row["intent"] for row in rows]
    print("Intent distribution:", dict(Counter(intents)))

    for row in rows[:2]:
        print("Example:", row["text"], "→", row["intent"])

    assert all("text" in row and "intent" in row for row in rows)

print("\n===== EXPECTED INTENTS =====")
print(settings.models.intent_classes)

print("\n===== DATASET INTENTS =====")
all_intents = set()

for split in ["train", "val", "test"]:
    with (data_dir / f"{split}.jsonl").open("r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                all_intents.add(json.loads(line)["intent"])

print(sorted(all_intents))

assert all_intents == set(settings.models.intent_classes)

print("\n✅ CELL PASSED — Transformer dataset and intent labels verified")
```

```python
# evaluating queries and retrieval
print("===== RETRIEVAL EVALUATION =====")

from src.retrieval.retriever import Retriever

eval_queries = [
    {
        "query": "How can I reset my password?",
        "relevant_document": "account_security.md"
    },
    {
        "query": "I was charged twice for the same payment.",
        "relevant_document": "billing_and_invoices.md"
    },
    {
        "query": "Can I get a refund for my subscription?",
        "relevant_document": "cancellation_and_refunds.md"
    },
    {
        "query": "Where can I track my package?",
        "relevant_document": "shipping_and_delivery.md"
    },
    {
        "query": "What is the weather today?",
        "relevant_document": None
    }
]

def evaluate_retrieval(threshold):
    retriever = Retriever(
        embedder=embedder,
        store=store
    )

    results = []

    for item in eval_queries:
        retrieved = retriever.search(
            item["query"],
            top_k=3,
            threshold=threshold
        )

        documents = [r["document"] for r in retrieved]
        relevant = item["relevant_document"]

        if relevant is None:
            recall = 1.0 if len(documents) == 0 else 0.0
            precision = 1.0 if len(documents) == 0 else 0.0
            rr = 1.0 if len(documents) == 0 else 0.0
        else:
            hit_positions = [
                i + 1
                for i, doc in enumerate(documents)
                if doc == relevant
            ]

            recall = 1.0 if hit_positions else 0.0
            precision = (
                sum(doc == relevant for doc in documents) / len(documents)
                if documents else 0.0
            )
            rr = 1.0 / hit_positions[0] if hit_positions else 0.0

        results.append({
            "query": item["query"],
            "relevant": relevant,
            "retrieved": documents,
            "recall": recall,
            "precision": precision,
            "rr": rr
        })

    return results


for threshold in [0.45, 0.50, 0.55]:
    results = evaluate_retrieval(threshold)

    recall_at_3 = sum(r["recall"] for r in results) / len(results)
    precision_at_3 = sum(r["precision"] for r in results) / len(results)
    mrr = sum(r["rr"] for r in results) / len(results)

    print(f"\nThreshold: {threshold}")
    print(f"Recall@3:    {recall_at_3:.4f}")
    print(f"Precision@3: {precision_at_3:.4f}")
    print(f"MRR:         {mrr:.4f}")

print("\n===== RETRIEVAL EVALUATION FINISHED =====")
print("✅ CELL COMPLETED")
```
